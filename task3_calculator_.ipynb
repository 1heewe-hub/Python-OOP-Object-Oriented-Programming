```python
class calculator:
    def add (self,a,b):
       return a + b
    def multiply(self,a,b):
      return a * b
    def subtract(self,a: float,b:int):
      return a - b
    def remainder(self,a,b):
      return a % b
    def divide(self,a,b):
      return a // b
    def divide2(self,a,b):
      return a / b
calc = calculator()
print("add =" , calc.add(5,3))
print("multiply=" ,calc.multiply(4,7))
print("subtract=" ,calc.subtract(5,3))
print("remainder=" ,calc.remainder(8.5,3))
print ("divide=" ,calc.divide(8,3))
print("divide2-" ,calc.divide2(7,3) )
```

```python
class Circle:
    def __init__(self, radius, pi=3.14):
        self.pi = pi
        self.radius = radius

    def area(self):
        return float(self.pi * (self.radius ** 2))

c1 = Circle(radius=5, pi=3.14)

print("Area =", c1.area())
```
